Given a list of integers, find the second-largest unique element without using sort() or sorted()

```python
nums = [10, 5, 20, 8, 20, 15]
nums = set(nums)

nums = list(nums)
greatest = nums[0]

for num in nums:
    if num> greatest:
        greatest = num

nums.remove(greatest)


greatest = nums[0]
for num in nums:
    if num > greatest:
        greatest = num

print(f'THis {greatest} is your second greatest number')
```

Q3. Find the frequency of elements

Count how many times each element appears in a list using a dictionary.

nums = [1, 2, 2, 3, 3, 3, 4, 4]

Expected output:
{1: 1, 2: 2, 3: 3, 4: 2}

```python
nums = [1, 2, 2, 3, 3, 3, 4, 4]

# Initialize an empty dictionary (using a safe name)
frequency_dict = {}

# Loop through each number directly
for num in nums:
    if num in frequency_dict:
        frequency_dict[num] += 1  # Increment the count if it exists
    else:
        frequency_dict[num] = 1   # Initialize the count to 1 if it's new

print(frequency_dict)
# Output: {1: 1, 2: 2, 3: 3, 4: 2}
```

Q6. Find the missing number


You are given a list containing numbers from 1 to N, but exactly one number is missing. Find that number.


nums = [1, 2, 4, 5, 6, 7]

n = 7

Expected output:3

```python
nums = [1,2,4,5,6,7]
n = 7

sum_of_n_numbers = n*(n+1)//2

actual_sum = sum(nums)

missing_number = sum_of_n_numbers - actual_sum

print(missing_number)
```

Q7. Find pairs with a given sum


Given a list and a target value, find all unique pairs whose sum equals the target.

nums = [2, 7, 11, 15, 3, 6, 4, 5]

target = 9

Expected output:[(2, 7), (3, 6), (4, 5)]

```python
nums = [2, 7, 11, 15, 3, 6, 4, 5]
target = 9

seen_dict = {}
mylist = []

for item in nums:
    remain = target - item
    
    # Check if the complement is already a key in our dictionary
    if remain in seen_dict:
        mylist.append((remain, item))
    
    # Store the current item in the dictionary
    seen_dict[item] = True

print(mylist)  
```
