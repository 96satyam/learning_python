Question 1 — Move all zeros to the end
Topic: Lists, loops, conditions, indexing, in-place logic.

Given a list of integers, move all zeros to the end of the list while maintaining the relative order of the non-zero elements.

You must not create a new list.

Example 1:
numbers = [0, 1, 0, 3, 12]Output:[1, 3, 12, 0, 0]



Example 2:
numbers = [0, 0, 5, 0, 2, 8]Output:[5, 2, 8, 0, 0, 0]



Example 3:
numbers = [1, 2, 3, 4]Output:[1, 2, 3, 4]



Constraints
- Don't create another list.
- Don't use sort() or sorted().
- Preserve the order of non-zero elements.
- Try to solve it with efficient logic rather than repeatedly removing elements.

```python
mylist = [1, 2, 3, 4]

write_position  = 0


for i in range(len(mylist)):
    if mylist[i]  != 0:
        mylist[write_position], mylist[i]  = mylist[i], mylist[write_position]

        write_position += 1

print(mylist)
```

Question 2 — Check whether two strings are anagrams
Topic: Strings, dictionaries, loops, conditions, frequency counting.

Write a Python program to check whether two strings are anagrams of each other.

Two strings are anagrams if they contain exactly the same characters with the same frequencies, but possibly in a different order.

Example 1:
str1 = "listen"str2 = "silent"Output:True



Example 2:
str1 = "hello"str2 = "world"Output:False



Example 3:
str1 = "aabbc"str2 = "abcab"Output:True



Constraints
- Don't use sorted().
- Don't use external libraries.
- Solve using Python logic and data structures you have studied.
- Try to achieve O(n) time complexity.

```python
str1 = "hello"
str2 = "silent"

dict1 = {}
dict2 = {}


for char in str1:
    if char in dict1:
        dict1[char] += 1
    else:
        dict1[char] = 1

for char in str2:
    if char in dict2:
        dict2[char] += 1
    else:
        dict2[char] = 1

#print(dict1)
#print(dict2)


if dict1 == dict2:
    print("Both sctrings are anagram")
else:
    print("Both Strings are not anagram")
```
