
Question 1. Find the second-largest distinct element in a list
Interview concept: Lists, loops, conditions, comparison operators.

Write a Python program to find the second-largest distinct number in a given list without using the sort() or sorted() functions.

Example 1:
numbers = [10, 5, 20, 8, 20, 15]Output:15



Example 2:
numbers = [10, 10, 10, 5, 5]Output:5



Requirements:
- Don't use sort() or sorted().
- Don't simply remove duplicates and sort.
- Handle duplicate numbers correctly.
- Consider what happens if a second-largest distinct element doesn't exist.

```python
numbers = [10,5,20,8,20,15]
#numbers = [10,10,10,10,10]

largest = float('-inf')

second_largest = float('-inf')

# now loop through each element in the list
#case 1 -------------> if num is greater than largest then new largest is num and second largest is old largest
# case2--------------> if num is less than largest also not equal to largest then second largest is num
# case3--------------> edge case when the second largest is still -inf then means no second largest is in list


for num in numbers:
    if num > largest:

        second_largest = largest
        largest = num

    elif num > second_largest and num != largest:
        second_largest = num


if second_largest == float('-inf'):
    print("there is no second largest element")

print(f"largest element is {largest}")
print(f"Second largest element is {second_largest}")
```

Question 2. Find the first non-repeating character in a string
Interview concept: Strings, dictionaries, loops, character frequency.

Write a Python program to find the first character in a string that appears only once.

Example 1:
text = "aabbcdeeff"Output:c



Example 2:
text = "programming"Output:p



Requirements:
- Maintain the original order of characters.
- Use Python logic rather than external libraries.
- Handle strings where every character is repeated.

```python
text = 'aabbcdeeff'

count = {}

for char in text:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1

for char in text:
    if count[char] == 1:
        print(f'this is the character {char}')
        break
```
