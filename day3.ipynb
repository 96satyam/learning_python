
Question 1 — Find the missing number
You are given a list of numbers from 1 to n, but one number is missing. Find the missing number.

The list contains no duplicates.

Example 1:
numbers = [1, 2, 3, 5, 6]Output:4



Example 2:
numbers = [2, 3, 4, 5, 7, 8, 9]Output:6



Example 3:
numbers = [1, 2, 4, 5, 6, 7]Output:3



Constraints
- Don't use sort() or sorted().
- Don't use external libraries.
- The numbers are from 1 to n.
- Exactly one number is missing.
- Try to solve it efficiently without nested loops.
Bonus: Can you solve it using only one loop and O(1) extra space?

```python
numbers = [1,2,4,5,6,7]

last_number = numbers[-1]


# total sum of all numbers from 1 to last number

total_sum = last_number * (last_number + 1) / 2 

# sum of all numbers of a list

list_sum  = sum(numbers)

# so the missing number is 

missing_number = total_sum - list_sum

if missing_number > 0:
    print(f"missing number is {missing_number}")

else:
    print("There is no missing number")
```

Question 2 — Find the most frequent element in a list
Given a list of integers, find the element that appears most frequently.

If two or more elements have the same maximum frequency, return any one of them.

Example 1:
numbers = [1, 2, 2, 3, 3, 3, 4]Output:3



Example 2:
numbers = [5, 5, 2, 2, 2, 5, 5]Output:5



Example 3:
numbers = [10, 20, 10, 30, 10, 20]Output:10



Constraints
- Use a dictionary to count frequencies.
- Don't use Counter.
- Don't use max() to directly find the answer.
- Use loops and conditions to determine the most frequent element.
Bonus: Print both the most frequent element and its frequency.

```python
numbers = [5,5,2,2,2,5,5]

my_dict = {}

for item in numbers:
    if item in my_dict:
        my_dict[item] += 1

    else:
        my_dict[item] = 1

print(my_dict)

max_value = 0
max_key = 0

for key,value in my_dict.items():

    if value > max_value:
        max_value = value
        max_key = key


print(f"maximum key in dictionary is  {max_key}")

```
